```python
# Code snippet 1
# Setting the dataset URL
url = 'https://raw.githubusercontent.com/nvamsimohan/DallasDSA/refs/heads/main/2023%20Texas%20High%20Schools%20Scores.csv'

# Display the url
url
```

```python
# Code snippet 2
# Importing Pandas package for data processing
import pandas
```

```python
# Code snippet 3
# Importing the dataset from the URL using pandas package
data = pandas.read_csv(url)

# Display the data
data
```

```python
# Code snippet 4
# Displaying the dataset size(rows and columns)
data.shape
```

```python
# Code snippet 5
# Displaying dataset columns
data.columns
```

```python
# Code snippet 6
# Displaying descriptive stats for the numerical variables
data.describe()
```

```python
# Code snippet 7
# Importing missingno package to display missing data proportions
import missingno as msno
```

```python
# Code snippet 8
# Displaying missing data proportion
msno.bar(data)
```

```python
# Code snippet 9
# Dropping  'Home Prices Ranking', 'Community Health Ranking' because of missing more than 20% data
# Quantitative Data Preparation
data80 = data.drop(['Home Prices Ranking', 'Community Health Ranking'], axis=1)

# Display the data80 columns
data80.columns
```

```python
# Code snippet 10
# Displaying data80 size/shape
data80.shape
```

```python
# Code snippet 11
# Setting the datatype for Zip as str
data80['Zip'] = data80['Zip'].astype(str)
```

```python
# Code snippet 12
# Dropping noise variables - 'School', 'School URL',  'District URL',  'City URL','Phone'
# Qualitative Data Preparation
predictors = data80.drop(['School', 'School URL',  'District URL',  'City URL','Phone'], axis=1)

# Display the predictors
predictors.columns
```

```python
# Code snippet 13
# Displaying predictors size/shape
predictors.shape
```

```python
# Code snippet 14
# Importing relevant statistical tests/modules from Dython
from dython.nominal import associations
```

```python
# Code snippet 15
# Running the associations function on predictor variables
assoc = associations(predictors)
```

```python
# Code snippet 16
# Storing the association values in a matrix
assoc_matrix = assoc['corr']
```

```python
# Code snippet 17
# Setting the target variable
target = 'Standard Score'
```

```python
# Code snippet 18
# Listing the most impacting/associated variables with the target variable
assoc_matrix[target].abs().sort_values(ascending=False)
```
