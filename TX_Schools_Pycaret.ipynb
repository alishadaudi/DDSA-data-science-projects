```python
# Code snippet 2
# Importing regression components from pycaret
from pycaret.regression import *
```

```python
# Code snippet 3
# Importing the dataset
import pandas as pd
data = pd.read_csv('https://raw.githubusercontent.com/nvamsimohan/DallasDSA/refs/heads/main/2023%20Texas%20High%20Schools%20Scores.csv')

# Display the data
data
```

```python
# Code snippet 4
# Displaying dataset columns
data.columns
```

```python
# Code snippet 5
# Setting the target variable
y = 'Standard Score'
```

```python
# Code snippet 6
# Setting the ignored columns list
ignored_cols = ['School', 'School URL', 'District URL', 'City URL',  'Phone',
                'Home Prices Ranking', 'Community Health Ranking']
```

```python
# Code snippet 7
# Setting the categorical columns list
cat_cols = ['District', 'City', 'Zip', 'County', 'Is Title I', 'Is Charter', 'Is Magnet',
      'Is Virtual']
```

```python
# Code snippet 8
# Setting the numerical columns list
num_cols  = ['Number Students', 'Number Full-time Teachers',
      'Student/Teacher Ratio', 'Percent Free/Disc Lunch',
      'Percent Homes Rented', 'Percent of Unemployment',
      'Percent Parents with Masters and above education',
      'Percent parents with less than High School education',
      'Percent Home Incomes with less than Median income']
```

```python
# Code snippet 9
# Setting/configuring ML setup with pycaret
regression_setup = setup(data,
                        target = y,
                        ignore_features = ignored_cols,
                        categorical_features = cat_cols,
                        numeric_features = num_cols)
```

```python
# Code snippet 10
# Invoking ML algorithms
compare_models()
```

```python
# Code snippet 11
# Performing cross validation (running the best algorithm 10 times)
best_model = create_model('huber')
```

```python
# Code snippet 12
# Importing the client dataset(for which the target values/std score needs to be predicted)
client_data = pd.read_csv('https://raw.githubusercontent.com/nvamsimohan/DallasDSA/refs/heads/main/TX%20High%20Schools%20-%20Client%20Data.csv')


client_data
```

```python
# Code snippet 13
# Predicting standard scores for the client dataset
predictions = predict_model(best_model, client_data)

# Exporting predictions to a CSV file
predictions.to_csv('TX Std Scores predictions.csv')
```
